### 3. Realizar a limpeza e o pré-processamento necessários.

```python
# Removendo colunas com valores ausentes
df = df.drop(columns=['formacao_mandante', 'formacao_visitante', 'tecnico_mandante', 'tecnico_visitante'])
df.head(10)
```

```python
# Removendo caracteres desnecessários
df["arena"] = df["arena"].str.replace("\xa0", " ", regex=False).str.strip()
df.head()
```

```python
# Convertendo a coluna data para datetime
df["data"] = pd.to_datetime(df["data"], dayfirst=True, errors="coerce")
```

```python
# Criando coluna com o ano
df["ano"] = df["data"].dt.year
df.head(20)
```

```python
# Criando df_2023 para analisar dados de 2023
df_2023 = df[df["ano"] == 2023].copy()
df_2023.tail()
```

```python
# Criando coluna resultado
df_2023["resultado"] = np.select(
    [df_2023["vencedor"] == df_2023["mandante"], df_2023["vencedor"] == df_2023["visitante"]],
    ["Mandante", "Visitante"],
    default="Empate"
)

df_2023.head()
```

### 4. Criar, no mínimo, três visualizações diferentes.

#### Visualização 1 - Resultados de jogos com vencedores mandantes, vencedores visitantes e empates

```python
resultado_counts = df_2023["resultado"].value_counts()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=resultado_counts.index,
    y=resultado_counts.values
)

plt.title("Resultados das partidas 2023")
plt.xlabel("Resultado")
plt.ylabel("Quantidade de partidas")
plt.tight_layout()
plt.show()
```

#### Visualização 2 - Times com mais vitórias em 2023

```python
vitorias_2023 = (
    df_2023[df_2023["vencedor"] != "-"]["vencedor"].value_counts().head(10)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=vitorias_2023.values,
    y=vitorias_2023.index
)

plt.title("Top 10 times com mais vitórias em 2023")
plt.xlabel("Número de vitórias")
plt.ylabel("Time")
plt.tight_layout()
plt.show()
```

#### Visualização 3 - Distribuição de quantidade de gols por partida em 2023

```python
# Criando a coluna de gols totais no df_2023
df_2023["gols_total"] = (
    df_2023["mandante_Placar"] + df_2023["visitante_Placar"]
)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df_2023,
    x="gols_total",
    bins=10,
    kde=True
)

plt.title("Distribuição de gols por partida em 2023")
plt.xlabel("Gols na partida")
plt.ylabel("Quantidade de partidas")
plt.tight_layout()
plt.show()
```

### 5. Apresentar conclusões e insights obtidos com a análise.

##### Conclusão da visualização 1
A distribuição entre vitórias de mandantes, visitantes e empates mostra que o fator jogar em casa influencia significativamente o resultado da partida.

##### Conclusão da visualização 2
Esses 10 times que aparecem nessa lista com mais vitórias não é exatamente a ordem de finalizou o campeonato, mas este indicador nos mostra os times com maior capacidade de converter os jogos em três pontos.

Times com muitas vitórias como mandante podem estar aproveitando bem a pressão do apoio da torcida, enquanto aqueles com mais vitórias fora de casa demonstram que o time é mais resiliente e táticamente preparada.

##### Conclusão da visualização 3
A análise da distribuição de gols nos mostra que a maioria das partidas se concentram em placares baixos, entre 1 a 3 gols, o que indica um campeonato com times defensivamente mais equilibrados e maior dificuldade das equipes de converter oportunidades em gols.
